```python
import pandas as pd
import matplotlib.pyplot as plt
import onnxruntime as ort
import numpy as np
import joblib

# model_path = '/export2/obasit/ClusterLevelServing/vllm_profiler_logs/power_model/'
# est_tp2_tp4 = joblib.load(model_path + "tp2_tp4/unified_power_model_tp2_tp4.joblib")
# est_tp8 = joblib.load(model_path + "tp8/unified_power_model_tp8.joblib")['model']

model_path = '/export2/obasit/ClusterLevelServing/vllm_profiler_logs/latency_model/'
# est_tp2_tp4 = joblib.load(model_path + "lightgbm_tp2_tp4/unified_latency_model_tp2_tp4.joblib")
est_tp8 = joblib.load(model_path + "lightgbm_tp8/unified_latency_model_tp8.joblib")['model']

def infer_prefill(model, tokens, tp, freq):
    input_feed = {
            "prefill_batch_size": [np.log1p(len(tokens))],
            "prefill_input_len_sum": [np.log1p(np.sum(tokens))],
            "prefill_input_len_mean": [np.log1p(float(np.mean(tokens)))],
            "prefill_input_len_std": [np.log1p(float(np.std(tokens)))],
            "decode_batch_size": [np.log1p(0)],
            "decode_input_len_sum": [np.log1p(0)],
            "decode_input_len_mean": [np.log1p(0)],
            "decode_input_len_std": [np.log1p(0)],
            "tp_degree": [np.log1p(tp)],
            "freq_mhz": [np.log1p(freq)],
        }

    # otherwise assume sklearn pipeline
    pred = model.predict(pd.DataFrame(input_feed), n_jobs=2)
    return float(pred[0])
```
